```python
import sys
import os
import shutil
import subprocess

# Dynamic check for Google Colab to securely append drive path
try:
    from google import colab
    # Only add the Drive path if running in Google Colab
    sys.path.append('/content/drive/MyDrive/Colab')
except ImportError:
    # If running in standard Jupyter Notebook / JupyterLab, use current directory
    if os.getcwd() not in sys.path:
        sys.path.append(os.getcwd())

# Import the pi_setup module
try:
    import Jpi_setup
    # Run the installation of pi and other tools
    Jpi_setup.install_all()
except ImportError:
    print("⚠️ Could not import 'pi_setup'. Ensure 'pi_setup.py' is placed in your current working directory.")

# Programmatically check tmux status
tmux_installed = shutil.which('tmux') is not None
tmux_running = False

if tmux_installed:
    try:
        # Check if tmux has any active sessions
        result = subprocess.run(['tmux', 'ls'], capture_output=True, text=True)
        if result.returncode == 0:
            tmux_running = True
    except Exception:
        pass

print("\n" + "="*100)
print("🎉 INSTALLATION COMPLETE!")
print("👉 Next Steps inside your terminal or notebook environment:")
print("1. Run: source ~/.bashrc               (To refresh your system paths)")
print("2. Run: pi                              (To launch the Pi AI CLI interface)")
print("3. Type: /login                         (Inside the Pi session to choose your model/provider)")

print("\n💡 Terminal Multiplexer (tmux) Status:")
if not tmux_installed:
    print("   - tmux is NOT installed. Please check your installation logs above.")
elif tmux_running:
    print("   - ✅ tmux is already running on this system!")
    print("   - Create a new terminal window: Press 'Ctrl-b' followed by 'c'")
    print("   - Switch between windows:       Press 'Ctrl-b' followed by a number '0-N'")
else:
    print("   - ℹ️ tmux is installed but NO sessions are currently running.")
    print("   - Start a new tmux session by running: tmux")
    print("   - Once inside tmux, you can use 'Ctrl-b c' to create windows and 'Ctrl-b 0-N' to switch.")
print("="*100)
```
