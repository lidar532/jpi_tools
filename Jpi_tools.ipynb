## 🛠️ JPI Tools Terminal Installation & Usage Guide

Follow these step-by-step instructions to initialize and configure the `pi` environment.

### 🚀 Step 1: Open Your Terminal in Colab
To run interactive terminal programs like `tmux` or the interactive `pi` interface, you need a terminal window:
1. Click on the **Terminal** icon on the left-hand sidebar menu in Google Colab (the console icon `>_` at the bottom left).
2. This opens a terminal pane at the bottom of your screen.

---

### 📦 Step 2: Run the Setup Script
In the Colab terminal pane, execute the installation script from the cloned repository. This script handles compiling/installing dependencies, setting up execution paths, and putting `pi` and other helpers in your system space:
```bash
bash /content/jpi_tools/setup_pi.sh
```

---

### 🔄 Step 3: Refresh Your Path Environment
Once the setup is complete, you must tell your current terminal session to read the newly added executable paths. Refresh your bash configuration:
```bash
source ~/.bashrc
```

---

### 🔐 Step 4: Launch and Log In
1. **Start the interactive interface** by running:
   ```bash
   pi
   ```
2. **Log in or choose your provider**: Inside the active `pi` session, type the command below to choose your model/provider and configure your API key:
   ```bash
   /login
   ```

---

### 💡 Tip: Using Pi Directly inside Notebook Cells
If you prefer running a quick query without switching to the terminal, you can run `pi` non-interactively using the `--print` (or `-p`) flag inside any notebook code cell:
```python
!pi --print "Explain quantum computing in one sentence."
```
